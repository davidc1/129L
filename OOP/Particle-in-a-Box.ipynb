```python
import matplotlib.patches as patches
```

```python
fig, ax = plt.subplots()
rect = patches.Rectangle((0, 0), 5, 5,\
                         linewidth=2, edgecolor='k', facecolor='none')

# reset position
p.setposition(2.5,2.5)
p.setvelocity(1,0)

# move until we hit a wall
while (p.inbox() == True):
    
    #ax.cla()
    # Add the patch to the Axes
    ax.add_patch(rect)

    
    ax.plot(p.x,p.y,'ro',markersize=3)
    
    
    p.update(0.03)
    
    ax.set_ylim([-1,6])
    ax.set_xlim([-1,6])

    display(fig)
    clear_output(wait=True)
    time.sleep(0.1)
    
print(p)
```
