### Import Particle Class and set up a Particle

```python
import particles
```

```python
p0 = particles.Particle(idx=0,g=-0.2)
p0.setposition(2.5,2.5)
p0.setvelocity(normal(),normal())
p1 = particles.Particle(idx=1,g=-0.2)
p1.setposition(2.5,2.5)
p1.setvelocity(normal(),normal())
p2 = particles.Particle(idx=2,g=-0.2)
p2.setposition(2.5,2.5)
p2.setvelocity(normal(),normal())
p3 = particles.Particle(idx=3,g=-0.2)
p3.setposition(2.5,2.5)
p3.setvelocity(normal(),normal())
p4 = particles.Particle(idx=4,g=-0.2)
p4.setposition(2.5,2.5)
p4.setvelocity(normal(),normal())
```

```python
p_v = [p0,p1,p2,p3,p4]
```

```python
import time
from IPython.display import display, clear_output
```

```python
import matplotlib.patches as patches
```

```python
fig, ax = plt.subplots()
rect = patches.Rectangle((0, 0), 5, 5,\
                         linewidth=2, edgecolor='k', facecolor='none')

# reset position
for p in p_v:
    p.setposition(2.5,2.5)
    p.setvelocity(normal(),normal())

# move until we hit a wall
while (p.inbox() == True):
    
    ax.cla()
    # Add the patch to the Axes
    ax.add_patch(rect)

    for p in p_v:
        ax.plot(p.x,p.y,'o',markersize=5)
        p.update(0.03)
    
    ax.set_ylim([-1,6])
    ax.set_xlim([-1,6])

    display(fig)
    clear_output(wait=True)
    time.sleep(0.1)
    
#print(p)
```
