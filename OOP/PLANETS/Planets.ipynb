### Import Planet Class and set up a Sun + Earth

```python
import planet
```

```python
d = 149597870700. # average distance from Sun to Earth [meters]
v = 29780. # average velocity of Earth around Sun [m/s]
Me = 5.97e24 # Kg
Ms = 1.99e30 # Kg
G = 6.67*(10**(-11))
Earth = planet.Planet(d,0,0,0,v,0,Me,1,G,0)
Sun = planet.Planet(0,0,0,0,0,0,Ms,1,G,1)
```

```python
D = Earth.calculate_distance(Sun)
print(d)

Earth.update_force(Sun)
```

```python
print('Earth starts at [%.02f, %.02f, %.02f] AUs from Sun'%(Earth.x/d,Earth.y/d,Earth.z/d))
print('Earth velocity is [%.02f, %.02f, %.02f] (frac of 29,780 m/s)'%(Earth.vx/v,Earth.vy/v,Earth.vz/v))
print('Acceleration is [%g, %g, %g] meters/sec'%(Earth.ax,Earth.ay,Earth.az))
print('Earth-Sun force is %g Newtons'%Earth.f)

F0 = Earth.f
```

```python
DT = 36000 # seconds
```

```python
Ex_v = []
Ey_v = []
Ez_v = []
Sx_v = []
Sy_v = []
Sz_v = []

dist_v = []
force_v = []

time_v = []

TOTTIME = 0

planet_v = [Earth,Sun]

for N in range(100000):
    
    Ex_v.append(Earth.x)
    Ey_v.append(Earth.y)
    Ez_v.append(Earth.z)
    
    Sx_v.append(Sun.x)
    Sy_v.append(Sun.y)
    Sz_v.append(Sun.z)
    
    Earth.update(DT,planet_v)
    
    dist_v.append( np.sqrt((Earth.x-Sun.x)**2 + (Earth.y-Sun.y)**2 + (Earth.z-Sun.z)**2) )
    force_v.append( Earth.f )
    
    time_v.append(DT*N)
    
    TOTTIME += DT
    
print('Total time elapsed: %g [days]'%(TOTTIME/(3600.*24)))
```

```python
fig = plt.figure(figsize=(6,6))
plt.plot(np.array(time_v)/(3600.*24),np.array(dist_v)/d)
plt.ylabel('Earth-Sun Distance [AU]')
plt.xlabel('Time [days]')
plt.show()
```

```python
fig = plt.figure(figsize=(6,6))
plt.plot(np.array(time_v)/(3600.*24),np.array(force_v)/F0)
plt.xlabel('Time [days]')
plt.ylabel('Force [frac. of initial]')
plt.show()
```

```python
fig = plt.figure(figsize=(6,6))

plt.plot(np.array(Ex_v)/d,np.array(Ey_v)/d,'k--')
plt.xlabel('x coordinate [AU]')
plt.ylabel('y coodinate [AU]')
plt.show()
```
