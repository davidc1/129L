### Import Planet Class and set up a 3-Body system that solves the stable "Figure 8" Orbit

```python
import planet
```

```python
x1, y1, z1 =  0.97000436,-0.24308753,0.0
x2, y2, z2 = -0.97000436,0.24308753,0.0
x3, y3, z3 =  0.0,0.0,0.0
```

```python
vx1, vy1, vz1 =  -0.93240737/2.,-0.86473146/2.,0.0
vx2, vy2, vz2 =  -0.93240737/2.,-0.86473146/2.,0.
vx3, vy3, vz3 = 0.93240737,0.86473146,0.
```

```python
M1 = 1.
M2 = 1.
M3 = 1.
G = 1.
P1 = planet.Planet(x1,y1,z1,vx1,vy1,vz1,M1,1,G,0)
P2 = planet.Planet(x2,y2,z2,vx2,vy2,vz2,M2,1,G,0)
P3 = planet.Planet(x3,y3,z3,vx3,vy3,vz3,M3,1,G,0)
```

```python
DT = 0.001 # second
```

```python
P1_x_v = []
P1_y_v = []
P1_z_v = []

P2_x_v = []
P2_y_v = []
P2_z_v = []

P3_x_v = []
P3_y_v = []
P3_z_v = []

time_v = []

TOTTIME = 0

planets_v = [P1,P2,P3]

for N in range(100000):
    
    P1_x_v.append(P1.x)
    P1_y_v.append(P1.y)
    P1_z_v.append(P1.z)
    
    P2_x_v.append(P2.x)
    P2_y_v.append(P2.y)
    P2_z_v.append(P2.z)
    
    P3_x_v.append(P3.x)
    P3_y_v.append(P3.y)
    P3_z_v.append(P3.z)
    
    #P1OLD = P1
    #P2OLD = P2
    #P3OLD = P3
    
    P1.update(DT,planets_v)
    P2.update(DT,planets_v)
    P3.update(DT,planets_v)
    
    
    time_v.append(DT*N)
    
    TOTTIME += DT
    
print('Total time elapsed: %g'%(TOTTIME))
```

```python
fig = plt.figure(figsize=(6,6))

plt.plot(np.array(P1_x_v),np.array(P1_y_v),'k.',label='Planet 1')
plt.plot(np.array(P2_x_v),np.array(P2_y_v),'r.',label='Planet 2')
plt.plot(np.array(P3_x_v),np.array(P3_y_v),'b.',label='Planet 3')
plt.xlabel('x coordinate')
plt.ylabel('y coodinate')
plt.legend(loc=1)
plt.show()
```
